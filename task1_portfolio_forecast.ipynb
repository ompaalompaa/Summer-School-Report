```python
import pandas as pd
```

```python
path = r"C:\Users\JOKER\Desktop\berlin\Fächer\Summer School\Power Markets\Data"
slp = pd.read_csv(f"{path}/slp_profiles.csv")
slp
```

```python
E_hb = 10000 * 3.5
E_gb = 200 * 30.0
E_lb = 20 * 100.0
```

```python
forecast = pd.DataFrame()
forecast["timestamp_utc"] = slp["timestamp_utc"]
forecast["timestamp_local"] = slp["timestamp_local"]
forecast["hb_forecast_mwh"] = E_hb * slp["hb_normalized_kwh"] / slp["hb_normalized_kwh"].sum()
forecast["gb_forecast_mwh"] = E_gb * slp["gb_normalized_kwh"] / slp["gb_normalized_kwh"].sum()
forecast["lb_forecast_mwh"] = E_lb * slp["lb_normalized_kwh"] / slp["lb_normalized_kwh"].sum()
forecast
```

```python
forecast["total_forecast_mwh"] = (forecast["hb_forecast_mwh"] + forecast["gb_forecast_mwh"] + forecast["lb_forecast_mwh"])
forecast
```

```python
print("Quarter-hours:", len(forecast))
print("HB annual MWh: expected", E_hb, "scaled", forecast["hb_forecast_mwh"].sum())
print("GB annual MWh: expected", E_gb, "scaled", forecast["gb_forecast_mwh"].sum())
print("LB annual MWh: expected", E_lb, "scaled", forecast["lb_forecast_mwh"].sum())
print("Total annual MWh: expected", E_hb + E_gb + E_lb, "scaled", forecast["total_forecast_mwh"].sum())
```

A quarter-hour holds 0.25 hours, so average MW = MWh / 0.25

```python
print("Average portfolio MW:", forecast["total_forecast_mwh"].mean() / 0.25)
print("Maximum quarter-hourly MWh:", forecast["total_forecast_mwh"].max())
print("Maximum average MW:", forecast["total_forecast_mwh"].max() / 0.25)
```
